```python
import PySimpleGUI as sg
import timeit
from unit_viewer_info import unitViewer,mockNode,mockExample,ExampleSet
```

```python
TOPFONT = ("Helvetica", 15)
FRAMEFONT = ("Helvetica", 15)
LISTFONT = ("Helvetica", 20)
BLANKFONT = 'any 20'
LISTENTRYSIZE = (23, 7)
SQUAREBOX = (2,10)
RECTANGULARBOX = (4,10)
CANVASWIDTH = 250
CAMVASHEIGHT = 235

ENABLEEVENTS = True
SUBMITCHAMGE = True
SQUARESIZE = 20
ALLTRAINING = ('0','1')



menu_def = [['&Viewer', ['&Update After',['Examples','Weight Updates','Progress Reports','Training and Testing','Never'],
                        '&Cell Size',['5','6','7','8','9','10','11','12','14','16','18','20','24','28','32','40'],
                        '&Cell Spacing',['0','1','2','3','4','6','8','10'],
                        '&Update','&Refresh','&Print...','&Close']],
            ['&Example Set',['Training Set', 'Testing Set']],
            ['&Procedure',['Test','Train']],
            ['&Value', ['Outputs and Targets', 'Outputs', 'Targets', 'Inputs', 'External Inputs', 'Output Derivatives',
                       'Input Derivatives', 'Gains', 'Link Weights', 'Link Derivs', 'Link Deltas']],
            ['&Palette',['Blue-Black-Red','Blue-Red-Yellow','Black-Gray-White','Hinton Diagram']]]



example_list_box = sg.Listbox(select_mode=sg.LISTBOX_SELECT_MODE_SINGLE,enable_events=True,values=ALLTRAINING, size=LISTENTRYSIZE,font=LISTFONT, key='_EXAMPLE_LIST_')
layout = [  [sg.Menu(menu_def, tearoff=True,font=TOPFONT)],

             [sg.Text('Event   ',font=TOPFONT), sg.Text('Example Time   ',font=TOPFONT),
              sg.Text('Event Time',font=TOPFONT),sg.Input(' ',size = (8,10), font = 'Any 20')],

             [sg.Input(' ',size = SQUAREBOX, font = BLANKFONT),sg.Text('/',font = BLANKFONT),sg.Input(' ',size = SQUAREBOX, font = BLANKFONT),sg.T('', font=BLANKFONT),
              sg.Input(' ',size = RECTANGULARBOX, font = BLANKFONT),sg.Text('/',font = BLANKFONT),sg.Input(' ',size = RECTANGULARBOX, font = BLANKFONT),sg.T('   ', font=BLANKFONT),
              sg.Input(' ',size = RECTANGULARBOX, font = BLANKFONT),sg.Text('/',font = BLANKFONT),sg.Input(' ',size = RECTANGULARBOX, font = BLANKFONT),
              sg.Input(' ',size = (8,10), font = 'Any 20'), sg.Slider(range=(1, 100), orientation='h', size=(34, 20), default_value=85)],

            [sg.Frame('Training Set', [[example_list_box]],font=FRAMEFONT),
            sg.Frame('Outputs and Targets',[[sg.Graph(canvas_size=(CANVASWIDTH, CAMVASHEIGHT), graph_bottom_left=(0, 0),
                                                        graph_top_right=(CANVASWIDTH, CAMVASHEIGHT),key="graph")]]),
            sg.Slider(range=(1, 100), orientation='v', size=(11, 20), default_value=50)]
        ]
# %%
window = sg.Window('unit viewer',layout, border_depth=0,use_default_focus=False)

window.Finalize()
graph = window.Element("graph")
rec_height = CAMVASHEIGHT/9
bias_v_coord = rec_height
input_v_coord = rec_height*3
hidden_v_coord = rec_height*5
output_v_coord = rec_height*7


example1 = mockExample([mockNode('red','0.52'),mockNode('red','0.52')],[mockNode('black','0.2'),mockNode('red','0.92'),mockNode('black','0.59')],
                       [mockNode('green','0.12','red'),mockNode('blue','0.7','green')])
example2 = mockExample([mockNode('black','0.97'),mockNode('black','0.91')],[mockNode('green','0.35'),mockNode('red','0.92'),mockNode('black','0.59')],
                       [mockNode('black','0.12','blue'),mockNode('blue','0.7','green')])
example_set = ExampleSet([example1,example2])
```

```python

def draw_layer(node_list,v_coord):
        h_coords = list(range(0,CANVASWIDTH,int(CANVASWIDTH/len(node_list))))
        for i in range(len(node_list)):
            h_coord = h_coords[i]
            graph.DrawRectangle((h_coord, v_coord), (h_coord + SQUARESIZE, v_coord + SQUARESIZE),
                                fill_color=(node_list[i]).color,line_color=(node_list[i]).border_color)

def draw_example(example):
    draw_layer(example.input_list,input_v_coord)
    draw_layer(example.hidden_list,hidden_v_coord)
    draw_layer(example.output_list,output_v_coord)
```

```python
while True:
    event, values = window.Read()
    if event is None or event == 'Exit':
        break
    if event == "_EXAMPLE_LIST_":
        idx = int(values['_EXAMPLE_LIST_'][0])
        draw_example(example_set.example_list[idx])
window.Close()
```
